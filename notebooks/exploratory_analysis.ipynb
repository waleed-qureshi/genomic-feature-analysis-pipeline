# Exploratory Genomic Analysis

This notebook provides an exploratory analysis of the results generated by the Genomic Feature Analysis Pipeline. We will examine sequence statistics and the distribution of genomic features (genes, CDS, and UTRs).

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os

plt.style.use('seaborn-v0_8-whitegrid')
print("Libraries imported successfully.")
```

## 1. Data Loading

We load the CSV files produced by the pipeline from the `results/` directory.

```python
results_dir = '../results'

try:
    seq_stats = pd.read_csv(os.path.join(results_dir, 'sequence_statistics.csv'))
    genes = pd.read_csv(os.path.join(results_dir, 'genes.csv'))
    cds = pd.read_csv(os.path.join(results_dir, 'cds.csv'))
    utr3 = pd.read_csv(os.path.join(results_dir, 'utr3.csv'))
    print("All result files loaded successfully.")
except FileNotFoundError as e:
    print(f"Error: {e}. Please run the main pipeline first.")
```

## 2. Sequence Statistics Analysis

Analyzing the overall properties of the genome assembly.

```python
print("Sequence Statistics Summary:")
display(seq_stats.head())

plt.figure(figsize=(6, 4))
plt.bar(seq_stats['sequence_id'], seq_stats['gc_percent'], color='skyblue', edgecolor='black')
plt.ylabel('GC Content (%)')
plt.title('GC Content per Sequence')
plt.xticks(rotation=45)
plt.show()
```

## 3. Feature Distribution

Comparing the lengths of genes and CDS to understand the coding efficiency.

```python
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.hist(genes['length'], bins=30, color='teal', edgecolor='black', alpha=0.7)
plt.title('Gene Length Distribution')
plt.xlabel('Length (bp)')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
plt.hist(cds['length'], bins=30, color='coral', edgecolor='black', alpha=0.7)
plt.title('CDS Length Distribution')
plt.xlabel('Length (bp)')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

## 4. 3' UTR Analysis

Examining the 3' UTR features if any were detected.

```python
if not utr3.empty:
    print(f"Found {len(utr3)} 3' UTR features.")
    display(utr3.head())
else:
    print("No 3' UTR features found in the current dataset.")
```

## 5. Summary Table

Quick overview of feature counts.

```python
summary = pd.DataFrame({
    'Feature': ['Genes', 'CDS', '3\' UTRs'],
    'Count': [len(genes), len(cds), len(utr3)]
})
display(summary)
```
